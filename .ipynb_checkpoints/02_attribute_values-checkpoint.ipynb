# 02 - Attribute values per product (item-attribute-value table)
**Purpose:** for every product assigned in notebook 01, select **exactly one** allowed value for **every** attribute that the configuration defines for its subcategory - with evidence, source, confidence and review status.

**Inputs:** `RES/01_product_subcategory_assignments.csv`, `Subcategories_Matrix_attributes.csv`  
**Outputs:** `RES/02_item_attribute_value.csv` (**main long table: one row per item x attribute**), `RES/02_product_attributes_wide.csv`, `RES/02_attribute_review_queue.csv`, `RES/dq_nb02.csv`  
**Packages:** pandas, numpy; `requests` only if `ENRICH_CONNOX = True`.

### How values are decided
1. **Product master first** - name, variant, colour, material and dimensions (mm) from `dwh_present_select.csv`.
2. **Brand heuristics** for design style / material where the text says nothing (documented in the evidence column).
3. **connox.de** (optional, switch below) to fill *low-confidence* values from the web product description.
4. If still unknown a **default guess** is stored (so every attribute has exactly one allowed value) and is marked `confidence = low`, `review_status = needs_review`, `value_is_default_guess = True`.

**Confidence:** `high` = stated directly in data (e.g. "Glasvase", "140 x 75 cm", volume in ml) - `medium` = derived from dimensions/keywords/brand/categories - `low` = weak signal or default.  
**Limits (please read):** the files contain **no images and no descriptions**, so "appearance" is inferred from the text only. Design style in particular is a brand/keyword heuristic and mostly `low`/`medium`.

```python
from pathlib import Path
import re, json, time, warnings, difflib, pandas as pd, numpy as np
warnings.filterwarnings('ignore', message='This pattern is interpreted as a regular expression')
pd.set_option('display.width', 200, 'display.max_colwidth', 80)

# ---------------- CONFIG (all assumptions are changeable here) ----------------
DATA_DIRS = [Path('SRC'), Path('.'), Path('..')/'SRC', Path('/mnt/data')]
OUT = Path('RES'); OUT.mkdir(exist_ok=True)
CONFIG_FILE = 'Subcategories_Matrix_attributes.csv'

ENRICH_CONNOX = False          # True = look up low-confidence products on connox.de (needs internet + requests)
ENRICH_MAX_PRODUCTS = 300      # safety cap per run
ENRICH_PAUSE_S = 1.0           # politeness delay between web requests

# Size thresholds (cm). Tables: rectangular LENGTH -> seats; round DIAMETER -> seats.
TABLE_RECT_CM = [(100, 'XS'), (150, 'S'), (190, 'M')]     # < first -> XS ... else L   (e.g. 160 cm -> M = 6 seats)
TABLE_ROUND_CM = [(90, 'XS'), (125, 'S'), (150, 'M')]
LOUNGE_WIDTH_FACTOR = 1.0      # seat width = overall width * factor (overall width from master 'Length'); 1.0 = use overall width
# Config thresholds from the attribute file (cm): lounge XS<65, S<75, M<85; shelf XS<60, S<100, M<180, L<250; lamp XS<20, S<40, M<80, L<150
LOUNGE_CM = [(65, 'XS'), (75, 'S'), (85, 'M')]
SHELF_CM = [(60, 'XS'), (100, 'S'), (180, 'M'), (250, 'L')]
LAMP_CM = [(20, 'XS'), (40, 'S'), (80, 'M'), (150, 'L')]
CUP_ML = [(100, 'XS'), (200, 'S'), (350, 'M')]
CUP_SHAPE_FACTOR = 0.6         # volume estimate from outer dimensions: pi/4*d^2*h*factor (used only as a low-confidence fallback)
```

## 1. Load assignments and attribute configuration

```python
def find_file(name):
    for d in DATA_DIRS:
        if (d/name).exists(): return d/name
    raise FileNotFoundError(name)
def read_csv_robust(path, **kw):
    for enc in ('utf-8-sig', 'cp1252', 'latin1'):
        try: return pd.read_csv(path, encoding=enc, **kw)
        except UnicodeDecodeError: continue
dq = []
def log(check, severity, detail, item_id=None, name=None, subcat=None):
    dq.append(dict(notebook='02', check=check, severity=severity, ItemId=item_id, product_name=name, subcategory=subcat, detail=detail))

cfg = read_csv_robust(find_file(CONFIG_FILE), dtype=str)
cfg['allowed'] = cfg['Values'].str.split(';').apply(lambda l: [v.strip() for v in l if v.strip()])
ATTR = {c: list(zip(g.Attribute, g.allowed)) for c, g in cfg.groupby('Category', sort=False)}
A = read_csv_robust(OUT/'01_product_subcategory_assignments.csv', dtype={'ItemId': str, 'SKU': str})
for c in ['Variant', 'ColorName', 'Material']: A[c] = A[c].where(A[c].notna(), None)
print(A.shape); A.config_subcategory.value_counts()
```

## 2. Text and colour helpers
Everything is matched on lower-cased text built from *Name (without brand) + Variant + Material + ColorName*. Units in the master are **mm**.

```python
def opt(allowed, *keys):
    """pick the allowed value whose label contains the first matching key (case-insensitive) - keeps labels exactly as configured"""
    for k in keys:
        for v in allowed:
            if (v.lower().startswith(k.lower()) if k.endswith(':') else k.lower() in v.lower()): return v
    raise KeyError(f'{keys} not in {allowed}')

def num(s): return float(str(s).replace(',', '.'))
def thr(value, table, last):                      # threshold table -> size label
    for lim, lab in table:
        if value < lim: return lab
    return last

COLOR = {   # family -> regex (German + English stems; no left boundary so compounds like 'hellblau' match)
 'white': r'wei(ß|ss)|white|creme|cream|elfenbein|ivory|bone|off-?white|snow|salt|chalk|pearl|perl|milk|alabaster|vanilla|marshmallow|nude|linen|leinen',
 'beige': r'beige|sand|\boat|hafer|greige|champagne|taupe|mushroom|putty|stone|natur',
 'grey':  r'grau|grey|gray|beton|concrete|pewter|moonbeam|mirage',
 'dark':  r'schwarz|black|anthrazit|anthracite|graphit|graphite|charcoal|\bink\b|onyx|slate|\bcoal\b|carbon|dunkel|dark|espresso',
 'metal': r'gold|silber|silver|messing|brass|kupfer|copper|chrom|chrome|bronze|nickel|edelstahl|stahl|steel|alumin|\balu\b',
 'clear': r'klar|transparent|clear|durchsichtig|glasklar|crystal|kristall|rauch|smoke',
 'earth': r'braun|brown|terracotta|terrakotta|ocker|ochre|senf|mustard|camel|cognac|rost|rust|caramel|\bclay\b|coffee|kaffee|chocolate|schoko|toffee|zimt|cinnamon|mokka|mocha|chestnut|honig|honey|hazel|walnuss',
 'cool':  r'blau|blue|gr(ü|ue)n|green|petrol|mint|t(ü|ue)rkis|turquoise|teal|oliv|olive|sage|salbei|navy|marine|aqua|lagoon|ocean|pacific|moss|khaki|forest|emerald|smaragd|jade|eukalyptus|\bsky\b|denim|cobalt|kobalt|cyan',
 'bold':  r'\brot|hellrot|dunkelrot|\bred\b|rosa|pink|orange|gelb|yellow|lila|violett|violet|purple|bordeaux|coral|korall|rubin|ruby|fuchsia|magenta|burgundy|cherry|kirsch|lemon|zitrone|tomato|peach|apricot|mehrfarbig|multi|bunt|\bmix\b|muster|print|gestreift|stripe|dots|punkte|floral|blumen|flieder|lavend|blush|mauve|plum|pflaume|sunshine',
}
LIGHT_PREFIX = re.compile(r'(hell|light|pale|cool|soft)\s*-?\s*$')

def colour_families(*texts):
    """ordered list of colour families named in the LAST comma-segment that contains a colour word (product colour is usually at the end of the name)."""
    for t in texts:
        if not t: continue
        t = re.sub(r'\(.*?\)', ' ', t.lower())
        for seg in reversed([s for s in re.split(r',', t) if s.strip()]):
            hits = []
            for fam, rx in COLOR.items():
                for m in re.finditer(rx, seg):
                    f = fam
                    if fam in ('grey', 'dark') and LIGHT_PREFIX.search(seg[:m.start()]): f = 'beige'    # hellgrau / light grey -> light neutral
                    hits.append((m.start(), f, m.group(0)))
            if hits:
                hits.sort(); fams = []
                for _, f, _w in hits:
                    if f not in fams: fams.append(f)
                return fams, seg.strip()
    return [], ''

def first_match(text, patterns):
    """earliest match among (label, regex) patterns -> (label, matched_text) or None"""
    best = None
    for lab, rx in patterns:
        m = re.search(rx, text)
        if m and (best is None or m.start() < best[1]): best = (lab, m.start(), m.group(0))
    return None if best is None else (best[0], best[2])

def R(value, source, evidence, rule, conf):
    return dict(value=value, source=source, evidence=evidence, rule=rule, confidence=conf)

def _s(x): return '' if x is None or (isinstance(x, float) and np.isnan(x)) else str(x)

class Ctx:
    def __init__(s, r):
        s.r = r; s.id = r.ItemId; s.sub = r.config_subcategory; s.shop_sub = r.ShopifySubCategory; s.main = r.ShopifyMainCategory
        name = _s(r.Name)
        s.brand = _s(r.brand).lower()
        s.name = re.sub(r'^.*?\s+-\s+', '', name, count=1)                # name without brand
        s.var, s.mat, s.col = _s(r.Variant), _s(r.Material), _s(r.ColorName)
        s.L, s.D, s.H = [None if pd.isna(x) or x <= 0 else float(x) for x in (r.Length, r.Depth, r.Height)]
        parts = []
        for p in (s.name, s.var, s.mat, s.col):
            if p and p.lower() not in ' | '.join(parts).lower(): parts.append(p)
        s.txt = ' | '.join(parts).lower()
        s.extra = ''; s.extra_url = None                                   # connox.de description (optional enrichment)
    @property
    def full(s): return s.txt + (' | ' + s.extra if s.extra else '')
    def src(s, base='dwh Name/Variant'): return base if not s.extra else base + ' + connox.de'
```

## 3. Design-style heuristic (shared by all subcategories)
Brand lists are **editable assumptions** (general design-world knowledge, not in the data). Product-name keywords override the brand.

```python
BRAND_STYLE = {
 'scandi':  ['hay','muuto','normann copenhagen','ferm living','menu','audo','string','moebe','frandsen','woud','northern','design house stockholm','iittala','form & refine','umage',
             'blomus','stelton','eva solo','aytm','&tradition','hübsch interior','lyngby','nuuck','by lassen','skagerak','karimoku','new works','vipp','bolia','ethnicraft','zone denmark','ro collection','rosendahl','nordal','cooee design','le klint','fritz hansen'],
 'classic': ['vitra','carl hansen','louis poulsen','knoll','artek','gubi','thonet','e15','cassina','georg jensen','royal copenhagen','rosenthal','arzberg','kpm','wmf','flos','zanotta','fermob','tolix','tom dixon','kay bojesen','pols potten'],
 'organic': ['muubs','hkliving','broste copenhagen','serax','house doctor','bloomingville','ellen & me','madam stoltz','by-boo','storus','cane-line','schönbuch','sika-design'],
 'playful': ['kartell','magis','seletti','oyoy','polspotten','design letters','marimekko','&klevering','bitten','moooi','ichendorf','donkey','rice','jellycat','sebra','liewood','bobo choses'],
 'geo':     ['101 copenhagen','tipton','tiptoe','pedestal','frost','nomess','kristina dam','sibast','mater','fatboy'],
}
KW_STYLE = [  # (key, regex) - name keywords override the brand
 ('organic', r'rattan|bast\b|geflecht|rope|seil|terrakotta|terracotta|handmade|handgemacht|raku|unregelm|woven|wicker|korb|bouclé|boucle|schafsfell|lamm|wolle|lehm|steinzeug|stoneware|reactive|glasur'),
 ('geo',     r'w(ü|ue)rfel|cube|geometr|square|quadrat|zylinder|cylinder|tube|bauhaus|\bgrid\b|block|modul|konfiguration|stacked|dreieck|triangle|disk|ring\b'),
 ('playful', r'bunt|smiley|bubble|wobbly|party|kids|kinder|mini bear|animal|tier|blume|flower|dots|punkte|stripe|gestreift|multi|mehrfarbig|printed|print\b|face|gesicht'),
 ('classic', r'classic|klassik|vintage|retro|wishbone|eames|chesterfield|royal|barock|1950|1960|panton|lc\d|grand repos|lounge chair|tulip|cesca|tizio|bauhaus'),
]
def style_key(c):
    hit = first_match(c.full.replace(c.brand, ''), KW_STYLE)
    if hit: return hit[0], 'dwh Name (keyword)', f'keyword "{hit[1]}"', 'medium'
    for k, brands in BRAND_STYLE.items():
        for b in brands:
            if c.brand == b:
                return k, 'brand heuristic', f'brand "{c.brand}" -> {k}', 'low'
    return 'scandi', 'default', 'no keyword or known brand; default = Scandinavian minimal (most common in assortment)', 'low'

STYLE_WORDS = {'scandi': ['scandinavian'], 'geo': ['geometric', 'sculptural'], 'classic': ['classic'], 'organic': ['organic'], 'playful': ['playful']}
def v_style(c, allowed):
    k, src, ev, conf = style_key(c)
    return R(opt(allowed, *STYLE_WORDS[k]), src, ev, 'style_' + ('keyword' if 'keyword' in src else 'brand' if 'brand' in src else 'default'), conf)
```

## 4. Attribute rules per subcategory
Each function returns one allowed value plus source, evidence, rule id and confidence.

```python
# ---------- VASE ----------
def vase_material(c, allowed):
    t = c.full
    m = first_match(t, [('Glass', r'glas|glass|kristall|crystal|\bklar\b|transparent|opal'), ('Porcelain', r'porzellan|porcelain|bone china'),
                        ('Ceramic', r'keramik|ceramic|steingut|steinzeug|stoneware|terrakotta|terracotta|ton\b|\bkrug\b|raku|earthenware'),
                        ('Metallic finish', r'metall|stahl|steel|messing|brass|alumin|kupfer|copper|chrom|bronze|gold|silber|silver|nickel')])
    kinds = {lab for lab, rx in [('Glass', r'glas|glass|kristall|crystal'), ('Metal', r'metall|stahl|steel|messing|brass|alumin|kupfer|copper|chrom|bronze|nickel'), ('Ceramic', r'keramik|ceramic|steingut|steinzeug|stoneware|porzellan|porcelain')] if re.search(rx, t)}
    if len(kinds) >= 2: return R(opt(allowed, 'mixed'), c.src(), f'several materials named: {sorted(kinds)}', 'vase_mat_mixed', 'medium')
    if m:
        lab, w = m; conf = 'high' if re.search(r'glas|porzellan|keramik|steingut|steinzeug|stahl|messing|alumin', w) else 'medium'
        return R(opt(allowed, lab), c.src(), f'keyword "{w}"', 'vase_mat_keyword', conf)
    fam, seg = colour_families(c.name, c.var, c.col)
    if 'clear' in fam: return R(opt(allowed, 'glass'), c.src(), f'colour "{seg}" implies transparent glass', 'vase_mat_clear', 'medium')
    return R(opt(allowed, 'ceramic'), 'default', 'no material word; ceramic is the most common vase material', 'vase_mat_default', 'low')

def vase_form(c, allowed):
    t = c.full; Hc = c.H / 10 if c.H else None
    wid = max([x for x in (c.L, c.D) if x] or [0]) / 10
    mH = re.search(r'\bh\s*(\d+(?:[.,]\d+)?)\s*cm', t)
    if mH: Hc, hsrc = num(mH.group(1)), 'name'
    else: hsrc = 'dwh Height'
    hit = first_match(t, [('Sculptural', r'skulptur|sculpt|objekt|object|\bart\b'), ('Floor', r'bodenvase|floor|boden-'), ('Cylinder', r'zylinder|cylinder'),
                          ('Tall', r'slim|schmal|bud\b|\btall\b|knospen|stiel|stem|\bflute\b'), ('Round', r'kugel|bauchig|\bround\b|\brund\b|bubble|sphere|\bball\b|bellied|\bkrug\b|belly|moon')])
    if hit:
        lab, w = hit
        v = {'Sculptural': 'sculptural', 'Floor': 'floor', 'Cylinder': 'large cylinder', 'Tall': 'tall', 'Round': 'round'}[lab]
        return R(opt(allowed, v), 'dwh Name (keyword)', f'keyword "{w}"', 'vase_form_keyword', 'medium')
    if Hc:
        ratio = Hc / wid if wid else None
        ev = f'height {Hc:.0f} cm ({hsrc})' + (f', width {wid:.0f} cm, ratio {ratio:.1f}' if wid else '')
        if Hc >= 50: return R(opt(allowed, 'floor'), f'dwh dimensions ({hsrc})', ev, 'vase_form_height', 'medium')
        if ratio and ratio >= 2.2 and Hc >= 20: return R(opt(allowed, 'tall'), 'dwh dimensions', ev, 'vase_form_ratio', 'medium')
        if ratio and ratio < 1.5: return R(opt(allowed, 'round'), 'dwh dimensions', ev, 'vase_form_ratio', 'medium')
        if Hc >= 28 and (not wid or wid >= 14): return R(opt(allowed, 'large cylinder'), 'dwh dimensions', ev, 'vase_form_ratio', 'low')
        return R(opt(allowed, 'tall' if Hc >= 22 else 'round'), 'dwh dimensions', ev, 'vase_form_ratio', 'low')
    return R(opt(allowed, 'round'), 'default', 'no height or shape keyword', 'vase_form_default', 'low')

def colour_map(c, allowed, mapping, default_val, rule, low_fams=()):
    fam, seg = colour_families(c.name, c.var, c.col, c.mat)
    if not fam and c.extra: fam, seg = colour_families(c.extra)
    if not fam: return R(default_val, 'default', 'no colour word found', rule + '_default', 'low')
    multi = len([f for f in fam if f not in ('metal',)]) >= 2 and not (set(fam) <= {'white', 'beige', 'grey'})
    f0 = fam[0]
    if 'bold' in fam and re.search(r'multi|mehrfarbig|bunt|\bmix\b|muster|print|gestreift|stripe', seg): f0, multi = 'bold', True
    val = mapping[f0]
    conf = 'high' if len(fam) == 1 and f0 not in low_fams else 'medium' if f0 not in low_fams else 'low'
    if multi and len(fam) > 1: conf = 'low' if conf != 'high' else 'medium'
    return R(val, c.src('dwh Name/Variant/ColorName'), f'colour segment "{seg}" -> {fam}', rule, conf)

def vase_colour(c, allowed):
    mp = dict(white=opt(allowed, 'white'), beige=opt(allowed, 'white'), grey=opt(allowed, 'white'), dark=opt(allowed, 'dark'), clear=opt(allowed, 'transparent'),
              metal=opt(allowed, 'metallic'), earth=opt(allowed, 'colourful'), cool=opt(allowed, 'colourful'), bold=opt(allowed, 'colourful'))
    return colour_map(c, allowed, mp, opt(allowed, 'white'), 'vase_colour', low_fams=('grey', 'earth'))

# ---------- DESIGN-ESSTISCH ----------
def parse_table_size(c):
    t = c.name.lower() + ' ' + c.var.lower()
    mr = re.search(r'[øo]\s*(\d{2,3}(?:[.,]\d+)?)', t)
    if mr and re.search(r'ø', t): return ('round', num(mr.group(1)), 'name "' + mr.group(0).strip() + '"', 'high')
    mx = re.search(r'(\d{2,3}(?:[.,]\d+)?)\s*x\s*(\d{2,3}(?:[.,]\d+)?)', t)
    if mx: return ('rect', max(num(mx.group(1)), num(mx.group(2))), 'name "' + mx.group(0) + ' cm"', 'high')
    ml = re.search(r'\b[lb]\s*(\d{2,3})\s*cm', t)
    if ml: return ('rect', num(ml.group(1)), 'name "' + ml.group(0) + '"', 'high')
    if c.L and c.D: return ('rect', max(c.L, c.D) / 10, f'dwh Length {c.L:.0f} mm x Depth {c.D:.0f} mm', 'medium')
    return None

def table_size(c, allowed):
    p = parse_table_size(c)
    if not p: return R(opt(allowed, 'M:'), 'default', 'no dimensions in name or master', 'table_size_default', 'low')
    kind, cm, ev, conf = p
    lab = thr(cm, TABLE_ROUND_CM if kind == 'round' else TABLE_RECT_CM, 'L')
    return R(opt(allowed, lab + ':'), 'dwh Name' if 'name' in ev else 'dwh dimensions', f'{kind} {cm:.0f} cm ({ev}) -> {lab}', 'table_size_' + kind, 'medium' if conf == 'high' else 'low')

def table_top(c, allowed):
    t = c.full
    top = re.split(r'gestell|untergestell|\bfu(ß|ss)|beine|\bbase\b|unterbau|kufen|sockel', t)[0]          # text before the base description
    top = top if len(top) > 8 else t
    hit = first_match(top, [('stone', r'marmor|marble|stein\b|granit|travertin|terrazzo|keramik|ceramic|beton|concrete|onyx|quarz|sintered|dekton'),
                            ('glass', r'glas|glass'), ('metal', r'metall|stahl|steel|alumin|messing|brass|edelstahl|kupfer|chrom|eisen|iron'),
                            ('wood', r'eiche|oak|esche|ash\b|buche|beech|ahorn|maple|birke|birch|kiefer|pine|holz|wood|bambus|nuss|walnu|teak|mahagoni|furnier|veneer|laminat|linoleum|hpl|mdf|fenix|melamin|lack|thermo')])
    if hit and hit[0] == 'stone': return R(opt(allowed, 'stone'), c.src(), f'keyword "{hit[1]}" in top description', 'table_top_stone', 'high')
    if hit and hit[0] == 'glass': return R(opt(allowed, 'glass'), c.src(), f'keyword "{hit[1]}"', 'table_top_glass', 'high')
    if hit and hit[0] == 'metal': return R(opt(allowed, 'metal'), c.src(), f'keyword "{hit[1]}"', 'table_top_metal', 'medium')
    dark = re.search(r'walnu|nussbaum|r(ä|ae)uchereiche|smoked|wenge|mocca|mokka|schwarz|black|dunkel|dark|espresso|anthrazit|thermo|mahagoni|braun|brown|ink', top)
    if hit:
        return R(opt(allowed, 'dark wood' if dark else 'light wood'), c.src(), f'wood/laminate word "{hit[1]}"' + (f', dark word "{dark.group(0)}"' if dark else ', no dark word'), 'table_top_wood', 'medium')
    fam, seg = colour_families(c.name, c.var)
    if fam and fam[0] == 'dark': return R(opt(allowed, 'dark wood'), c.src(), f'only colour "{seg}"', 'table_top_colour', 'low')
    return R(opt(allowed, 'light wood'), 'default', 'no top material word; light wood is the most common', 'table_top_default', 'low')

def table_func(c, allowed):
    t = c.full
    hit = first_match(t, [('extendable', r'ausziehbar|ausziehtisch|extendable|extension|verl(ä|ae)ngerbar|\bausz|butterfly'), ('foldable', r'klapptisch|klappbar|foldable|faltbar|klappen|fold'),
                          ('console', r'konsole|console|wandtisch|wandmontage|wall'), ('modular', r'modular|modul\b|kombinierbar|combin')])
    if hit: return R(opt(allowed, {'extendable': 'extendable', 'foldable': 'foldable', 'console': 'wall', 'modular': 'modular'}[hit[0]]), c.src(), f'keyword "{hit[1]}"', 'table_func_keyword', 'high')
    return R(opt(allowed, 'fixed'), 'default (absence of keywords)', 'no extension/folding/modular wording in name or variant', 'table_func_absence', 'medium')

# ---------- LOUNGESESSEL ----------
def lounge_width(c, allowed):
    t = c.name.lower() + ' ' + c.var.lower()
    m = re.search(r'\b[bw]\s*(\d{2,3})\s*cm', t)
    if m: cm, ev, src, conf = num(m.group(1)) * LOUNGE_WIDTH_FACTOR, f'name "{m.group(0)}"', 'dwh Name', 'medium'
    elif c.L: cm, ev, src, conf = c.L / 10 * LOUNGE_WIDTH_FACTOR, f'master Length {c.L:.0f} mm (overall width used as seat width)', 'dwh dimensions', 'medium'
    else: return R(opt(allowed, 'M:'), 'default', 'no width available', 'lounge_width_default', 'low')
    lab = thr(cm, LOUNGE_CM, 'L')
    return R(opt(allowed, lab + ':'), src, f'{cm:.0f} cm -> {lab}; {ev}', 'lounge_width', conf)

def lounge_material(c, allowed):
    t = c.full
    if re.search(r'kunstleder|leatherette|faux', t): return R(opt(allowed, 'leather'), c.src(), 'artificial leather treated as Leather (no better option)', 'lounge_mat_fauxleather', 'low')
    if re.search(r'leder|leather|nappa|anilin|premium f\b|cognac', t): return R(opt(allowed, 'leather'), c.src(), 'keyword leather/Leder', 'lounge_mat_leather', 'high')
    fab = re.search(r'stoff|bouclé|boucle|kvadrat|samt|velvet|cord\b|wolle|wool|schafsfell|sheepskin|leinen|linen|textil|webstoff|polster|sunbrella|fabric|tonus|hallingdal|remix|vidar|steelcut|divina|ruskin|rohi|sandow|gabriel|mollis|clay|bezug|kissen|upholster', t)
    plastic = re.search(r'kunststoff|plastic|polypropylen|polycarbonat|shell|schale', t)
    wood = re.search(r'eiche|oak|esche|ash\b|buche|nuss|walnu|teak|holz|wood|birke|bambus|mahagoni', t)
    metal = re.search(r'outdoor|alumin|stahl|steel|metall|rattan|geflecht|rope|seil|teak', t)
    fam, seg = colour_families(c.name, c.var, c.col)
    neutral, colour = opt(allowed, 'neutral fabric'), opt(allowed, 'colour or pattern')
    def fabcol(conf, src_ev):
        if not fam: return R(neutral, c.src(), src_ev + '; no colour word -> neutral', 'lounge_mat_fabric_nocolour', 'low')
        if fam[0] in ('white', 'beige', 'grey', 'dark'): return R(neutral, c.src(), f'{src_ev}; colour "{seg}" neutral', 'lounge_mat_fabric_neutral', conf if fam[0] != 'dark' else 'low')
        return R(colour, c.src(), f'{src_ev}; colour "{seg}" -> {fam[0]}', 'lounge_mat_fabric_colour', conf)
    if fab: return fabcol('high', f'fabric word "{fab.group(0)}"')
    if plastic: return R(opt(allowed, 'metal/outdoor'), c.src(), f'plastic shell word "{plastic.group(0)}" - no matching option, nearest = Metal/outdoor frame', 'lounge_mat_plastic', 'low')
    if metal and not wood: return R(opt(allowed, 'metal/outdoor'), c.src(), f'keyword "{metal.group(0)}"', 'lounge_mat_metal', 'medium')
    if wood and not fam: return R(opt(allowed, 'wood frame'), c.src(), f'wood word "{wood.group(0)}"', 'lounge_mat_wood', 'low')
    return fabcol('low', 'no material word, upholstered armchair assumed')

def lounge_func(c, allowed):
    t = c.full
    hit = first_match(t, [('ottoman', r'(sessel|chair|lounge)[^|]{0,15}(und|mit|&|\+)\s*(ottoman|hocker|fu(ß|ss)hocker|footstool)|set mit (ottoman|hocker)|ottoman set'),
                          ('swivel', r'drehbar|swivel|drehgestell|dreh(fu|st)|\bturn\b|rotat'), ('rocking', r'schaukel|rocking|\brocker\b|gleit|glider|wippe'),
                          ('recline', r'verstellbar|recliner|liegefunktion|relax|neigbar|recline|kopfst(ü|ue)tze verstellbar|adjustable')])
    if hit: return R(opt(allowed, {'ottoman': 'ottoman', 'swivel': 'swivel', 'rocking': 'rocking', 'recline': 'recline'}[hit[0]]), c.src(), f'keyword "{hit[1]}"', 'lounge_func_keyword', 'high' if hit[0] != 'recline' else 'medium')
    return R(opt(allowed, 'fixed frame'), 'default (absence of keywords)', 'no mechanism wording; mechanisms (e.g. swivel bases) are often not named in the title - verify', 'lounge_func_absence', 'medium')

# ---------- REGALSYSTEME ----------
def shelf_width(c, allowed):
    t = c.name.lower() + ' ' + c.var.lower()
    mh = re.search(r'\bh\s*(\d{2,3})\s*x\s*(\d{2,3})', t)           # 'H 180 x 84 cm'  -> width = second number
    mx = re.search(r'(\d{2,3}(?:[.,]\d+)?)\s*x\s*(\d{2,3}(?:[.,]\d+)?)(?:\s*x\s*(\d{2,3}(?:[.,]\d+)?))?\s*cm', t)
    mb = re.search(r'\b[bl]\s*(\d{2,3})\s*cm', t)
    if mh: cm, ev, src, conf = num(mh.group(2)), f'name "{mh.group(0)}" (H x width)', 'dwh Name', 'high'
    elif mb: cm, ev, src, conf = num(mb.group(1)), f'name "{mb.group(0)}"', 'dwh Name', 'high'
    elif mx: cm, ev, src, conf = num(mx.group(1)), f'name "{mx.group(0)}" (first number = width, assumption)', 'dwh Name', 'medium'
    elif c.L: cm, ev, src, conf = c.L / 10, f'master Length {c.L:.0f} mm', 'dwh dimensions', 'medium'
    else: return R(opt(allowed, 'S:'), 'default', 'no width available', 'shelf_width_default', 'low')
    lab = thr(cm, SHELF_CM, 'XL')
    if re.search(r'konfiguration|\bset\b|doppel|double|\d\s*-?\s*(feld|bay)', t) and lab in ('M', 'L'):
        lab, ev, conf = 'XL', ev + ' + multi-bay wording', 'medium'
    return R(opt(allowed, lab + ':'), src, f'{cm:.0f} cm -> {lab}; {ev}', 'shelf_width', conf)

def shelf_material(c, allowed):
    t = c.full
    top = t
    cands = [('light wood', r'eiche|oak|esche|ash\b|buche|beech|birke|birch|kiefer|pine|ahorn|maple|bambus|bamboo|sperrholz|plywood|natur'),
             ('dark wood', r'walnu|nussbaum|r(ä|ae)uchereiche|smoked|wenge|mocca|mokka|dark oak|dunkle eiche|mahagoni|teak'),
             ('plastic', r'kunststoff|plastic|polypropylen|\babs\b|acryl')]
    fam, seg = colour_families(c.name, c.var, c.col)
    hit = first_match(top, cands)
    if hit and hit[0] == 'plastic': return R(opt(allowed, 'colourful or plastic'), c.src(), f'keyword "{hit[1]}"', 'shelf_mat_plastic', 'medium')
    colour_hit = None
    if fam:
        m = {'white': 'white', 'beige': 'white', 'grey': 'white', 'dark': 'black', 'metal': 'black', 'clear': 'white', 'earth': 'colourful', 'cool': 'colourful', 'bold': 'colourful'}[fam[0]]
        colour_hit = (m, seg)
    if hit:
        lab, w = hit
        conf = 'high' if not colour_hit or fam[0] in ('white', 'beige') or lab == 'dark wood' else 'medium'
        return R(opt(allowed, lab), c.src(), f'wood word "{w}"' + (f'; colour segment "{seg}"' if colour_hit else ''), 'shelf_mat_wood', conf)
    if colour_hit:
        v = {'white': opt(allowed, 'white'), 'black': opt(allowed, 'black'), 'colourful': opt(allowed, 'colourful')}[colour_hit[0]]
        return R(v, c.src('dwh Name/Variant/ColorName'), f'colour segment "{seg}" -> {fam}', 'shelf_mat_colour', 'medium' if len(fam) == 1 else 'low')
    return R(opt(allowed, 'white'), 'default', 'no material/colour word', 'shelf_mat_default', 'low')

def shelf_func(c, allowed):
    t = c.full
    hit = first_match(t, [('wall', r'wandregal|wandmontiert|wandmontage|floating|schwebe|wandmodul|h(ä|ae)ngeregal|wall shelf|wall mounted|wall shelving'), ('ladder', r'leiter|ladder|leaning|lehnregal|anlehn'),
                          ('enclosed', r'\bt(ü|ue)r|schublade|schrank|door|drawer|vitrine|cabinet|klappt|kommode|sideboard|container'),
                          ('modular', r'system|modul|konfiguration|erweiterung|add-on|zusatz|element|einlegeboden|\bboden|erg(ä|ae)nzung|combin|stacked|metallboden|ablage|podium'),
                          ('free', r'standregal|freistehend|b(ü|ue)cherregal|\bunit\b|\bregal\b')])
    key = {'wall': 'wall-mounted', 'ladder': 'ladder', 'enclosed': 'enclosed', 'modular': 'modular', 'free': 'freestanding'}
    if hit: return R(opt(allowed, key[hit[0]]), c.src(), f'keyword "{hit[1]}"', 'shelf_func_keyword', 'medium' if hit[0] in ('modular', 'free') else 'high')
    return R(opt(allowed, 'freestanding'), 'default', 'no functional keyword; freestanding single unit assumed', 'shelf_func_default', 'low')

# ---------- AKKULEUCHTEN ----------
def lamp_height(c, allowed):
    t = c.name.lower() + ' ' + c.var.lower()
    m = re.search(r'\bh(?:\u00f6he)?\s*(\d{1,3}(?:[.,]\d+)?)\s*cm', t)
    if m: cm, ev, src, conf = num(m.group(1)), f'name "{m.group(0)}"', 'dwh Name', 'high'
    elif c.H: cm, ev, src, conf = c.H / 10, f'master Height {c.H:.0f} mm', 'dwh dimensions', 'medium'
    else: return R(opt(allowed, 'S:'), 'default', 'no height available', 'lamp_height_default', 'low')
    lab = thr(cm, LAMP_CM, 'XL')
    return R(opt(allowed, lab + ':'), src, f'{cm:.1f} cm -> {lab}; {ev}', 'lamp_height', conf)

def lamp_material(c, allowed):
    t = c.full
    hit = first_match(t, [('ceramic', r'keramik|porzellan|ceramic|porcelain|steinzeug'), ('glass', r'glas|glass|opal|kristall'), ('fabric', r'stoff|textil|fabric|leinen|linen|plissee|schirm aus|stoffschirm'),
                          ('plastic', r'silikon|silicone|kunststoff|plastic|\babs\b|polycarbonat|\bpc\b|gummi|rubber|acryl|resin|harz'),
                          ('metal', r'metall|alumin|stahl|steel|messing|brass|chrom|gold|kupfer|copper|nickel|eisen|iron|bronze')])
    if hit: return R(opt(allowed, hit[0]), c.src(), f'keyword "{hit[1]}"', 'lamp_mat_keyword', 'high' if hit[1] not in ('opal', 'gold') else 'medium')
    if re.search(r'solar', t): return R(opt(allowed, 'plastic'), c.src(), 'solar lamps are typically plastic housings', 'lamp_mat_solar', 'low')
    return R(opt(allowed, 'metal'), 'default', 'no material word; metal is the most common housing material for portable lamps', 'lamp_mat_default', 'low')

def lamp_control(c, allowed):
    t = c.full
    hit = first_match(t, [('rgb', r'\brgb\b|farbwechsel|colou?r[- ]changing|mehrfarbig|multicolou?r'), ('tune', r'\btune\b|tunable|\bcct\b|farbtemperatur|warmwei(ß|ss) bis|warm to cool|einstellbare lichtfarbe'),
                          ('smart', r'smart|\bapp\b|bluetooth|fernbedienung|remote|wifi|wlan|alexa|hue'), ('dim', r'dimm|touch|dimmer|stufig|helligkeit|dimmable|dim\b')])
    if hit: return R(opt(allowed, {'rgb': 'colour-changing', 'tune': 'tunable', 'smart': 'smart', 'dim': 'dimmable'}[hit[0]]), c.src(), f'keyword "{hit[1]}"', 'lamp_ctrl_keyword', 'high')
    return R(opt(allowed, 'fixed warm'), 'default', 'no light-control wording in the master data; many portable lamps have dimming - needs verification (connox.de lookup helps)', 'lamp_ctrl_default', 'low')

OUTDOOR_BRANDS = ['fermob', 'cane-line', 'skagerak', 'ethimo', 'gandia blasco', 'tribù']
def lamp_env(c, allowed):
    t = c.full
    ip = re.search(r'\bip\s?(\d{2})\b', t)
    if ip: return R(opt(allowed, 'full outdoor' if int(ip.group(1)) >= 44 else 'covered'), c.src(), f'IP rating {ip.group(0)}', 'lamp_env_ip', 'high')
    if re.search(r'balkon|terrasse|covered|wettergesch', t): return R(opt(allowed, 'covered'), c.src(), 'balcony/terrace wording', 'lamp_env_covered', 'medium')
    if re.search(r'outdoor|au(ß|ss)en|garten|wetterfest|weatherproof|\bsolar\b', t) or c.main == 'Außenleuchten':
        why = 'Shopify main category Außenleuchten' if c.main == 'Außenleuchten' else 'outdoor/solar wording'
        return R(opt(allowed, 'full outdoor'), c.src('dwh Name/category'), why, 'lamp_env_outdoor', 'medium')
    if c.brand in OUTDOOR_BRANDS: return R(opt(allowed, 'covered'), 'brand heuristic', f'brand {c.brand} makes outdoor furniture/lighting; not confirmed for this item', 'lamp_env_brand', 'low')
    return R(opt(allowed, 'indoor'), 'default (absence of keywords)', 'no outdoor wording or IP rating; indoor category', 'lamp_env_absence', 'medium')

# ---------- TASSEN ----------
def cup_volume(c):
    t = (c.name + ' ' + c.var).lower()
    m = re.search(r'(\d+(?:[.,]\d+)?)\s*(ml|cl|l)\b', t)
    if not m: return None
    v = num(m.group(1)); ml = v if m.group(2) == 'ml' else v * 10 if m.group(2) == 'cl' else v * 1000
    return ml, m.group(0)

def cup_size(c, allowed):
    t = c.full
    mset = re.search(r'\b(\d+)\s*er[- ]?(set|pack)|\b(\d+)-tlg|\bduo\b|\bset of\b|\b(\d+)er\b', t)
    if mset: return R(opt(allowed, 'set'), c.src(), f'pack wording "{mset.group(0)}"', 'cup_size_set', 'high')
    v = cup_volume(c)
    if v:
        lab = thr(v[0], CUP_ML, 'L')
        return R(opt(allowed, lab + ':'), 'dwh Name', f'stated volume {v[1]} -> {lab}', 'cup_size_volume', 'high')
    if re.search(r'espresso|mokka|mocha', t) or c.shop_sub == 'Espressotassen': return R(opt(allowed, 'XS:'), c.src('dwh Name/category'), 'espresso cup', 'cup_size_espresso', 'medium')
    if re.search(r'cappuccino', t): return R(opt(allowed, 'S:'), c.src(), 'cappuccino cup', 'cup_size_cappuccino', 'medium')
    if re.search(r'\bgro(ß|ss)\b|\blarge\b|\bxl\b|jumbo|latte|schale', t): return R(opt(allowed, 'L:'), c.src(), 'large wording', 'cup_size_large', 'low')
    dims = [x for x in (c.L, c.D) if x]
    if dims and c.H:
        d = min(dims) / 10; ml = np.pi / 4 * d * d * (c.H / 10) * CUP_SHAPE_FACTOR
        lab = thr(ml, CUP_ML, 'L')
        return R(opt(allowed, lab + ':'), 'dwh dimensions (estimate)', f'est. volume ~{ml:.0f} ml from min(L,D)={d:.1f} cm, H={c.H/10:.1f} cm (packaging/handle may distort)', 'cup_size_dims', 'low')
    return R(opt(allowed, 'M:'), 'default', 'no volume or dimension', 'cup_size_default', 'low')

BRAND_CUP_MAT = {'fine porcelain': ['design letters','iittala','marimekko','rosenthal','arzberg','kpm','royal copenhagen','georg jensen','stelton','blomus','menu','audo','villeroy','wmf','by lassen','design house stockholm','nuuck'],
                 'stoneware': ['hkliving','bloomingville','broste copenhagen','muubs','house doctor','ferm living','&klevering','oyoy','hay','muuto','serax','aytm','madam stoltz','ro collection']}
def cup_material(c, allowed):
    t = c.full
    hit = first_match(t, [('fine porcelain', r'porzellan|porcelain|bone china|vitro'), ('stoneware', r'steinzeug|stoneware|keramik|ceramic|steingut|earthenware'), ('glass', r'glas|glass'),
                          ('enamel', r'emaille|enamel|edelstahl|metall|stahl|steel|alumin'), ('other', r'kunststoff|plastic|bambus|bamboo|melamin|holz|silikon|kork|cork|pla\b')])
    if hit: return R(opt(allowed, hit[0]), c.src(), f'keyword "{hit[1]}"', 'cup_mat_keyword', 'high')
    for k, bl in BRAND_CUP_MAT.items():
        if c.brand in bl: return R(opt(allowed, k), 'brand heuristic', f'brand "{c.brand}" mostly makes {k} tableware (not confirmed for this item)', 'cup_mat_brand', 'low')
    if re.search(r'untertasse|saucer', t): return R(opt(allowed, 'fine porcelain'), c.src(), 'cup with saucer set - typically porcelain', 'cup_mat_saucer', 'low')
    return R(opt(allowed, 'stoneware'), 'default', 'no material word; stoneware is the most common in this assortment', 'cup_mat_default', 'low')

def cup_colour(c, allowed):
    mp = dict(white=opt(allowed, 'neutral light'), beige=opt(allowed, 'earth'), earth=opt(allowed, 'earth'), grey=opt(allowed, 'dark'), dark=opt(allowed, 'dark'),
              cool=opt(allowed, 'cool'), bold=opt(allowed, 'bold'), clear=opt(allowed, 'neutral light'), metal=opt(allowed, 'dark'))
    return colour_map(c, allowed, mp, opt(allowed, 'neutral light'), 'cup_colour', low_fams=('clear', 'metal', 'beige'))

def cup_func(c, allowed):
    t = c.full
    hit = first_match(t, [('double', r'doppelwandig|double[- ]wall|thermo|isolier|insulated'), ('infuser', r'infuser|sieb|teesieb|filter|strainer'), ('lid', r'untertasse|deckel|\blid\b|saucer|mit unterteller'),
                          ('handleless', r'ohne henkel|handleless|henkellos'), ('handle', r'mit henkel|\bhenkel\b|handle')])
    key = {'double': 'double-walled', 'infuser': 'infuser', 'lid': 'lid or saucer', 'handleless': 'handleless', 'handle': 'standard handle'}
    if hit: return R(opt(allowed, key[hit[0]]), c.src(), f'keyword "{hit[1]}"', 'cup_func_keyword', 'high')
    if re.search(r'\bbecher\b', c.name.lower()): return R(opt(allowed, 'handleless'), c.src('dwh Name/category'), '"Becher" without "Henkel" - mugs without handle are common in this assortment; verify', 'cup_func_becher', 'low')
    return R(opt(allowed, 'standard handle'), 'default', '"Tasse" without other wording - standard handle assumed', 'cup_func_tasse', 'medium')

RULES = {
 ('Vase', 'Design style'): v_style, ('Vase', 'Material'): vase_material, ('Vase', 'Form & flower fit'): vase_form, ('Vase', 'Colour & finish'): vase_colour,
 ('Design-Esstisch', 'Design style'): v_style, ('Design-Esstisch', 'Size (seats)'): table_size, ('Design-Esstisch', 'Material & look (table top)'): table_top, ('Design-Esstisch', 'Functionality'): table_func,
 ('Loungesessel', 'Design style'): v_style, ('Loungesessel', 'Size (seat width)'): lounge_width, ('Loungesessel', 'Material & look'): lounge_material, ('Loungesessel', 'Functionality'): lounge_func,
 ('Regalsysteme', 'Design style'): v_style, ('Regalsysteme', 'Size (overall width)'): shelf_width, ('Regalsysteme', 'Material & look'): shelf_material, ('Regalsysteme', 'Functionality'): shelf_func,
 ('Akkuleuchten', 'Design style'): v_style, ('Akkuleuchten', 'Size (height)'): lamp_height, ('Akkuleuchten', 'Material & look'): lamp_material, ('Akkuleuchten', 'Light control'): lamp_control, ('Akkuleuchten', 'Environment'): lamp_env,
 ('Tassen', 'Design style'): v_style, ('Tassen', 'Size (volume)'): cup_size, ('Tassen', 'Material'): cup_material, ('Tassen', 'Colour & finish'): cup_colour, ('Tassen', 'Functionality'): cup_func,
}
missing_rules = [(c, a) for c, l in ATTR.items() for a, _ in l if (c, a) not in RULES]
for c, a in missing_rules: log('attribute_without_rule', 'high', f'No rule implemented for {c}/{a} - attribute will be UNDETERMINED', subcat=c)
print('attributes in config:', sum(len(v) for v in ATTR.values()), '| without rule:', missing_rules)
```

## 5. Optional connox.de enrichment
Off by default. When `ENRICH_CONNOX = True`, products that have at least one **low-confidence** attribute are looked up on connox.de (Shopify storefront search suggestions; fallback: first product link of the search page). The product description is appended to the text and the rules are re-run; a web-based value replaces the earlier one **only if its confidence is higher**, and the source is then recorded as `connox.de <url>`. Responses are cached in `RES/connox_cache.json`.  
*Note: this block needs internet and was written for the public Shopify endpoints; it could not be run in the offline environment where the notebooks were built, so check a few results manually.*

```python
CACHE_FILE = OUT/'connox_cache.json'
cache = json.loads(CACHE_FILE.read_text()) if CACHE_FILE.exists() else {}
def strip_html(h): return re.sub(r'\s+', ' ', re.sub(r'<[^>]+>', ' ', h or '')).strip()

def connox_lookup(query):
    import requests, urllib.parse
    if query in cache: return cache[query]
    base = 'https://www.connox.de'; hdr = {'User-Agent': 'Mozilla/5.0 (attribute-research notebook)'}
    res = None
    try:
        r = requests.get(base + '/search/suggest.json', params={'q': query, 'resources[type]': 'product', 'resources[limit]': 3}, headers=hdr, timeout=20)
        prods = r.json()['resources']['results']['products']
        if prods:
            best = max(prods, key=lambda p: difflib.SequenceMatcher(None, query.lower(), p.get('title', '').lower()).ratio())
            score = difflib.SequenceMatcher(None, query.lower(), best.get('title', '').lower()).ratio()
            res = dict(url=base + best['url'].split('?')[0], title=best.get('title'), text=strip_html(best.get('body')) + ' ' + ' '.join(best.get('tags') or []),
                       image=(best.get('image') or ''), score=round(score, 2))
    except Exception as e:
        res = dict(error=str(e)[:200])
    cache[query] = res; CACHE_FILE.write_text(json.dumps(cache, ensure_ascii=False)); time.sleep(ENRICH_PAUSE_S)
    return res
```

## 6. Run all rules - one value per item and attribute

```python
def run_rules(c):
    out = {}
    for attr, allowed in ATTR[c.sub]:
        f = RULES.get((c.sub, attr))
        if f is None:
            out[attr] = R('UNDETERMINED', 'none', 'no rule implemented', 'none', 'low'); continue
        try:
            res = f(c, allowed)
        except Exception as e:                                   # never lose a record: log and fall back
            res = R(allowed[0], 'error', f'rule error: {e}', 'error', 'low'); log('rule_error', 'high', f'{c.sub}/{attr}: {e}', c.id, c.r.Name, c.sub)
        out[attr] = res
    return out

ranks = {'low': 0, 'medium': 1, 'high': 2}
records, enriched = [], 0
for _, r in A.iterrows():
    c = Ctx(r)
    res = run_rules(c)
    if ENRICH_CONNOX and enriched < ENRICH_MAX_PRODUCTS and any(v['confidence'] == 'low' for v in res.values()):
        look = connox_lookup(f"{r.brand} {c.name.split(',')[0]}")
        enriched += 1
        if look and look.get('text') and look.get('score', 0) >= 0.5:
            c.extra, c.extra_url = look['text'].lower(), look['url']
            res2 = run_rules(c)
            for a in res:
                if ranks[res2[a]['confidence']] > ranks[res[a]['confidence']]:
                    res[a] = dict(res2[a], source='connox.de ' + look['url'], evidence=res2[a]['evidence'] + f" (web match \"{look['title']}\", score {look['score']})")
        elif look and look.get('error'): log('connox_lookup_failed', 'medium', look['error'], r.ItemId, r.Name, r.config_subcategory)
    for attr, allowed in ATTR[c.sub]:
        v = res[attr]
        valid = v['value'] in allowed
        if not valid: log('invalid_value_selected', 'high', f"{attr}: '{v['value']}' not in allowed list", r.ItemId, r.Name, c.sub)
        records.append(dict(ItemId=r.ItemId, SKU=r.SKU, product_name=r.Name, brand=r.brand, subcategory=c.sub, attribute=attr, selected_value=v['value'],
                            allowed_values=' ; '.join(allowed), source=v['source'], evidence=v['evidence'], rule_id=v['rule'], confidence=v['confidence'],
                            value_is_default_guess=v['source'].startswith('default'), value_valid=valid,
                            assignment_confidence=r.assignment_confidence, assignment_review_status=r.review_status))
iav = pd.DataFrame(records)
iav['review_status'] = np.select([iav.confidence == 'high', iav.confidence == 'medium'], ['confirmed_from_data', 'inferred_optional_review'], default='needs_review')
iav.loc[(iav.assignment_review_status == 'needs_review') & (iav.review_status != 'needs_review'), 'review_status'] = 'assignment_needs_review'
print(iav.shape, '| connox lookups:', enriched)
```

## 7. Integrity checks (exactly one valid value per item x attribute)

```python
exp = A.merge(cfg[['Category', 'Attribute']], left_on='config_subcategory', right_on='Category')[['ItemId', 'Attribute']].rename(columns={'Attribute': 'attribute'})
chk = exp.merge(iav[['ItemId', 'attribute']].assign(n=1).groupby(['ItemId', 'attribute']).n.sum().reset_index(), how='left', on=['ItemId', 'attribute'])
missing = chk[chk.n.isna()]; multi = chk[chk.n > 1]
print('expected item x attribute rows:', len(exp), '| missing:', len(missing), '| duplicates:', len(multi), '| invalid values:', (~iav.value_valid).sum())
assert len(missing) == 0 and len(multi) == 0 and iav.value_valid.all(), 'integrity problem - see log'
for sc, g in iav.groupby('subcategory'):
    n_low = (g.confidence == 'low').groupby(g.attribute).mean().round(2).to_dict()
    log('attribute_low_confidence_share', 'info', f'{sc}: share of low-confidence values per attribute: {n_low}', subcat=sc)
for _, r in iav[(iav.confidence == 'low') & iav.value_is_default_guess].groupby(['subcategory', 'attribute']).size().reset_index(name='n').iterrows():
    log('default_guess_values', 'medium', f'{r.n} values for {r.attribute} are pure default guesses (no evidence)', subcat=r.subcategory)
for _, r in A[A.ItemId.duplicated(keep=False)].iterrows(): log('duplicate_assignment', 'high', 'ItemId assigned more than once', r.ItemId, r.Name, r.config_subcategory)
pd.crosstab([iav.subcategory, iav.attribute], iav.confidence)
```
