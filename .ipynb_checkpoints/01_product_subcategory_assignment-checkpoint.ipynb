# 01 - Product-to-subcategory assignment
**Purpose:** read the subcategory configuration, load the product master (`dwh_present_select.csv`) and the sales item list, and assign every relevant product to exactly one configured subcategory.

**Inputs** (searched in `./SRC`, `.`, `../SRC`): `Subcategories_Matrix_attributes.csv`, `dwh_present_select.csv`, `1. Group_ Orders & sales.csv`  
**Outputs** (in `./RES`): `01_product_subcategory_assignments.csv`, `01_assignment_review.csv`, `dq_nb01.csv`  
**Packages:** pandas, numpy (see `environment.yml`). Source files are only read, never modified.

Key assumptions are in the *CONFIG* cell and can be changed there.

```python
from pathlib import Path
import re, warnings, pandas as pd, numpy as np
warnings.filterwarnings('ignore', message='This pattern is interpreted as a regular expression')
pd.set_option('display.width', 200, 'display.max_colwidth', 90)

# ---------------- CONFIG ----------------
DATA_DIRS = [Path('SRC'), Path('.'), Path('..')/'SRC', Path('/mnt/data')]
OUT = Path('RES'); OUT.mkdir(exist_ok=True)
CONFIG_FILE, PRODUCT_FILE, SALES_FILE = 'Subcategories_Matrix_attributes.csv', 'dwh_present_select.csv', '1. Group_ Orders & sales.csv'

# Mapping: config subcategory -> how it is found in the product master.
#  shopify_subs  : ShopifySubCategory values that ARE this subcategory (assumption, edit freely)
#  name_regex    : name pattern used (a) to rescue products with missing Shopify category ("'-")
#                  and (b) to list look-alike products from OTHER Shopify subcategories for review
#  component_regex: products that match are kept but flagged as 'component / accessory - review'
SUBCATEGORY_MAP = {
 'Vase':            dict(shopify_subs=['Vasen'],
                         name_regex=r'(?i)\b(deko-?|glas-?|boden-?)?vasen?\b|bodenvase|glasvase',
                         component_regex=None),
 'Design-Esstisch': dict(shopify_subs=['Esstische'],
                         name_regex=r'(?i)esstisch',
                         component_regex=r'(?i)tischplatte|tischbein|ansteckplatte|einlegeplatte|verl\u00e4ngerung'),
 'Loungesessel':    dict(shopify_subs=['Loungesessel'],
                         name_regex=r'(?i)lounge ?(chair|sessel)|loungesessel',
                         component_regex=r'(?i)^(?!.*(sessel|chair)).*(ottoman|hocker|kissen|bezug|h\u00fclle)'),
 'Regalsysteme':    dict(shopify_subs=['Regalsysteme'],
                         name_regex=r'(?i)regalsystem|shelving system',
                         component_regex=r'(?i)\b(metallboden|bodenleiter|podium|ablage|fachboden|einlegeboden|r\u00fcckwand|seitenwand|zubeh\u00f6r|halter)\b'),
 'Akkuleuchten':    dict(shopify_subs=['Akkuleuchten'],
                         name_regex=r'(?i)akku|portable|kabellos|cordless',
                         component_regex=r'(?i)ladestation|ladekabel|ladeschale|ersatz|zubeh\u00f6r|lampenschirm'),
 'Tassen':          dict(shopify_subs=['Tassen'],   # strict: 'Becher' and 'Espressotassen' are separate Shopify subcategories; add them here to include them
                         name_regex=r'(?i)\b(kaffee|tee|espresso|cappuccino)?tassen?\b|kaffeebecher|teebecher',
                         component_regex=r'(?i)^(?!.*(mit|und|inkl\.?) untertasse).*untertasse'),
}
ALLOW_NAME_RESCUE = False   # STRICT RULE: a product is assigned only if its ShopifySubCategory is listed in shopify_subs. True = also assign by name when Shopify category is missing
UNASSIGNED_MARKER = "'-"        # value used in the master for 'no Shopify category'
```

## 1. Load configuration and check it

```python
def find_file(name):
    for d in DATA_DIRS:
        if (d/name).exists(): return d/name
    raise FileNotFoundError(f'{name} not found in {DATA_DIRS}')
def read_csv_robust(path, **kw):
    for enc in ('utf-8-sig', 'cp1252', 'latin1'):
        try: return pd.read_csv(path, encoding=enc, **kw), enc
        except UnicodeDecodeError: continue
dq = []   # data-quality log: every issue is appended here and saved at the end
def log(check, severity, detail, item_id=None, name=None, subcat=None):
    dq.append(dict(notebook='01', check=check, severity=severity, ItemId=item_id, product_name=name, subcategory=subcat, detail=detail))

cfg, enc = read_csv_robust(find_file(CONFIG_FILE), dtype=str)
print('config encoding:', enc, cfg.shape)
cfg['allowed_values'] = cfg['Values'].str.split(';').apply(lambda l: [v.strip() for v in l if v.strip()])
# config checks
for (c, a), g in cfg.groupby(['Category', 'Attribute']):
    if len(g) > 1: log('config_duplicate_attribute', 'high', f'{c}/{a} defined {len(g)}x', subcat=c)
for _, r in cfg.iterrows():
    if len(r.allowed_values) < 2: log('config_attribute_few_values', 'medium', f'{r.Category}/{r.Attribute}: {r.allowed_values}', subcat=r.Category)
    if len(set(r.allowed_values)) != len(r.allowed_values): log('config_duplicate_values', 'medium', f'{r.Category}/{r.Attribute}', subcat=r.Category)
subcats = list(dict.fromkeys(cfg.Category))
for c in subcats:
    if c not in SUBCATEGORY_MAP: log('unmapped_config_subcategory', 'high', f'{c} has no entry in SUBCATEGORY_MAP - no products can be assigned', subcat=c)
cfg.groupby('Category').size().rename('n_attributes').to_frame()
```

## 2. Load product master and sales item list (read-only) and profile them

```python
prod, enc_p = read_csv_robust(find_file(PRODUCT_FILE), dtype=str)
sales, enc_s = read_csv_robust(find_file(SALES_FILE), dtype=str, keep_default_na=False)
sales.columns = ['ItemId', 'sales_name', 'sales_item_type', 'noi1', 'qty1', 'noi2', 'qty2', 'stock']
sales = sales[sales.ItemId != 'Total']            # drop the 'Total' summary row (re-used for reconciliation in notebook 03)
print('products', prod.shape, enc_p, '| sales items', sales.shape, enc_s)
raw_n = len(prod)

# --- duplicates of ItemId (kept visible: identical copies are collapsed, conflicting copies are logged) ---
dups = prod[prod.ItemId.duplicated(keep=False)].sort_values('ItemId')
for iid, g in dups.groupby('ItemId'):
    ident = g.drop_duplicates().shape[0] == 1
    diff_cols = [c for c in g.columns if g[c].nunique(dropna=False) > 1]
    log('duplicate_item_id', 'medium' if ident else 'high',
        ('identical copies' if ident else 'rows differ in: ' + ', '.join(diff_cols)) + f' ({len(g)} rows) - first row kept', iid, g.Name.iloc[0], g.ShopifySubCategory.iloc[0])
prod = prod.drop_duplicates('ItemId', keep='first').copy()

# --- placeholders: '0' in Variant/ColorName/Material and "'-" in categories mean 'no data' ---
for c in ['Variant', 'ColorName', 'Material']:
    prod[c] = prod[c].replace({'0': np.nan, '': np.nan})
for c in ['Length', 'Depth', 'Height', 'Weight']:
    prod[c] = pd.to_numeric(prod[c], errors='coerce')
for c in ['ShopifyTopCategory', 'ShopifyMainCategory', 'ShopifySubCategory']:
    prod[c] = prod[c].replace({'': np.nan})
prod['brand'] = prod.Name.str.extract(r'^(.*?)\s+-\s+')[0]
prod['in_sales_file'] = prod.ItemId.isin(sales.ItemId)
print('after de-dup:', len(prod), '| unassigned Shopify category:', (prod.ShopifySubCategory == UNASSIGNED_MARKER).sum(),
      '| no brand parsed:', prod.brand.isna().sum(), '| not in sales file:', (~prod.in_sales_file).sum())
log('product_master_summary', 'info', f'{raw_n} rows, {len(prod)} unique ItemId; {prod.Name.isna().sum()} without name; Material filled for {prod.Material.notna().sum()}, ColorName for {prod.ColorName.notna().sum()}, Variant for {prod.Variant.notna().sum()} rows')
log('no_image_or_description_columns', 'high', 'Product master has no image URL or description column; attributes can only come from Name/Variant/dimensions (+ optional connox.de lookup in notebook 02)')
log('no_sku_column', 'info', 'No separate SKU column; ItemId is used as SKU / product ID')
log('unknown_item_type_codes', 'info', 'ItemTypeId_CNX20 has codes ' + ', '.join(sorted(prod.ItemTypeId_CNX20.dropna().unique())) + ' - meaning not documented, kept as-is')
for c in ['Length', 'Depth', 'Height']:
    z = prod[c].fillna(-1).le(0).sum()
    log('dimension_missing_or_zero', 'info', f'{c}: {z} products with missing/zero value (round items often have Length=Depth=0)')
prod[['ItemId','Name','ShopifyTopCategory','ShopifyMainCategory','ShopifySubCategory']].head()
```

## 3. Match products to subcategories
Rules (in priority order): **A (the only assigning rule)** the product's Shopify subcategory is one of `shopify_subs` -> assigned (`high` confidence; `medium` if the Shopify main category is unusual);  
**B** Shopify category missing (`'-`) but name matches -> **not assigned** (unless `ALLOW_NAME_RESCUE = True`), listed in the review file;  
**C** name matches but product sits in a *different* Shopify subcategory -> **not assigned**, listed in the review file.
Products without an assignment are kept in the review output, never dropped.

```python
rows = []
for sc in subcats:
    m = SUBCATEGORY_MAP.get(sc)
    if not m: continue
    subs = m['shopify_subs']
    in_sub = prod.ShopifySubCategory.isin(subs)
    name_hit = prod.Name.fillna('').str.contains(m['name_regex'], regex=True)
    comp = prod.Name.fillna('').str.contains(m['component_regex'], regex=True) if m['component_regex'] else pd.Series(False, index=prod.index)
    # most common main category of the Shopify subcategory = 'expected' main category
    exp_main = prod[in_sub].ShopifyMainCategory.value_counts()
    exp_main = set(exp_main[exp_main / exp_main.sum() > 0.2].index)
    for kind, mask in [('A', in_sub), ('B', (prod.ShopifySubCategory == UNASSIGNED_MARKER) & name_hit), ('C', name_hit & ~in_sub & (prod.ShopifySubCategory != UNASSIGNED_MARKER))]:
        t = prod[mask].copy(); t['config_subcategory'] = sc; t['rule'] = kind; t['is_component'] = comp[mask].values
        t['main_ok'] = t.ShopifyMainCategory.isin(exp_main)
        rows.append(t)
cand = pd.concat(rows, ignore_index=True)
print(cand.groupby(['config_subcategory', 'rule']).size().unstack(fill_value=0))
```

```python
# ---- resolve: each ItemId gets at most ONE subcategory (priority A > B, then config order) ----
assign = cand[cand.rule.isin(['A', 'B'] if ALLOW_NAME_RESCUE else ['A'])].copy()
assign['prio'] = assign.rule.map({'A': 0, 'B': 1}); assign['ord'] = assign.config_subcategory.map({s: i for i, s in enumerate(subcats)})
assign = assign.sort_values(['ItemId', 'prio', 'ord'])
multi = assign[assign.ItemId.duplicated(keep=False)]
for iid, g in multi.groupby('ItemId'):
    log('conflicting_subcategory_match', 'high', 'matches ' + ', '.join(g.config_subcategory) + f' -> kept {g.config_subcategory.iloc[0]}', iid, g.Name.iloc[0], g.config_subcategory.iloc[0])
assign = assign.drop_duplicates('ItemId', keep='first').copy()

def basis(r):
    if r.rule == 'A': return 'Shopify subcategory = ' + r.ShopifySubCategory
    return "Shopify category missing ('-); matched by product name"
def conf(r):
    if r.rule == 'B': return 'low'
    return 'high' if r.main_ok else 'medium'
def reason(r):
    out = []
    if r.rule == 'B': out.append("Shopify category missing - assigned from name only")
    if r.rule == 'A' and not r.main_ok: out.append(f'unusual Shopify main category "{r.ShopifyMainCategory}" for this subcategory')
    if r.is_component: out.append('looks like a component / accessory, not a full product')
    if not r.in_sales_file: out.append('item not in sales file')
    return '; '.join(out)
assign['assignment_basis'] = assign.apply(basis, axis=1)
assign['assignment_confidence'] = assign.apply(conf, axis=1)
assign['review_reason'] = assign.apply(reason, axis=1)
assign['review_status'] = np.where(assign.review_reason.str.contains('component|missing|unusual'), 'needs_review', 'ok')
assign['SKU'] = assign.ItemId
cols = ['ItemId','SKU','Name','brand','ManufacturerId','config_subcategory','assignment_basis','assignment_confidence','review_status','review_reason',
        'ShopifyTopCategory','ShopifyMainCategory','ShopifySubCategory','Length','Depth','Height','Weight','Variant','ColorName','Material','ItemTypeId_CNX20','in_sales_file']
assign = assign[cols].sort_values(['config_subcategory', 'ItemId']).reset_index(drop=True)
assign.to_csv(OUT/'01_product_subcategory_assignments.csv', index=False, encoding='utf-8-sig')
print(assign.groupby(['config_subcategory','assignment_confidence']).size().unstack(fill_value=0))
print(assign.review_status.value_counts().to_dict())
```

## 4. Review output - nothing is silently dropped
Includes: look-alike products in other Shopify subcategories (rule C), low-confidence/component assignments, and (separate file) sales-file items missing from the master that match a subcategory by name. Master products that were not assigned to any configured subcategory (counted, not listed).

```python
review = []
# C: look-alikes in other Shopify subcategories (not assigned)
c = cand[cand.rule.isin(['B', 'C'])]
c = c[~c.ItemId.isin(assign.ItemId)]
for _, r in c.iterrows():
    review.append(dict(ItemId=r.ItemId, Name=r.Name, source='product master', candidate_subcategory=r.config_subcategory, current_category=f'{r.ShopifyMainCategory} / {r.ShopifySubCategory}',
                       review_type=('no_assignment_shopify_category_missing_name_match' if r.rule == 'B' else 'no_assignment_name_match_in_other_shopify_subcategory'), explanation=('Shopify category missing; name suggests this subcategory but strict rule requires a matching Shopify subcategory' if r.rule == 'B' else 'Name matches but Shopify subcategory differs; not assigned (edit SUBCATEGORY_MAP to include)')))
# components / low-confidence assignments
for _, r in assign[assign.review_status == 'needs_review'].iterrows():
    review.append(dict(ItemId=r.ItemId, Name=r.Name, source='product master', candidate_subcategory=r.config_subcategory, current_category=f'{r.ShopifyMainCategory} / {r.ShopifySubCategory}',
                       review_type='assigned_needs_review', explanation=r.review_reason))
# sales-only items matching by name -> separate file (can be thousands of rows; mostly delisted/old items)
known = set(prod.ItemId)
so = sales[~sales.ItemId.isin(known)]
so_rows = []
for sc in subcats:
    m = SUBCATEGORY_MAP.get(sc)
    if not m: continue
    h = so[so.sales_name.str.contains(m['name_regex'], regex=True)]
    so_rows.append(h.assign(candidate_subcategory=sc, review_type='sales_item_not_in_master_name_match',
        explanation='Exists only in the sales file (no category/dimensions); name suggests this subcategory; NOT assigned.')[['ItemId','sales_name','sales_item_type','candidate_subcategory','review_type','explanation']])
sales_only = pd.concat(so_rows, ignore_index=True)
n_sales_only_hits = len(sales_only)
sales_only.to_csv(OUT/'01_sales_only_name_candidates.csv', index=False, encoding='utf-8-sig')
print('sales-only name candidates:', sales_only.groupby('candidate_subcategory').size().to_dict())
review = pd.DataFrame(review)
# subcategories with no / few products
for sc in subcats:
    n = (assign.config_subcategory == sc).sum()
    if n == 0: log('subcategory_without_products', 'high', f'{sc}: no products assigned', subcat=sc)
unassigned_master = prod[~prod.ItemId.isin(assign.ItemId)]
log('master_products_outside_config', 'info', f'{len(unassigned_master)} of {len(prod)} master products belong to none of the {len(subcats)} configured subcategories (expected - config covers only some subcategories)')
log('sales_items_not_in_master', 'medium', f'{len(so)} of {len(sales)} sales-file items are not in the product master ({n_sales_only_hits} of them match a configured subcategory by name; listed in review)')
log('master_items_not_in_sales', 'info', f'{(~prod.in_sales_file).sum()} master items have no row in the sales file')
review.to_csv(OUT/'01_assignment_review.csv', index=False, encoding='utf-8-sig')
pd.DataFrame(dq).to_csv(OUT/'dq_nb01.csv', index=False, encoding='utf-8-sig')
print(review.review_type.value_counts() if len(review) else 'no review rows')
review.head(10)
```
