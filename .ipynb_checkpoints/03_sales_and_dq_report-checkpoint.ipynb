# 03 - Sales summaries and data-quality report
**Purpose:** summarise sales per product, subcategory and category; check the requested periods; consolidate the data-quality log of all notebooks into one exception report and bundle everything in one Excel workbook.

**Inputs:** `1. Group_ Orders & sales.csv`, `dwh_present_select.csv`, outputs of notebooks 01 and 02  
**Outputs:** `05_sales_*.csv`, `06_data_quality_exception_report.csv`, `Product_Attribute_Report.xlsx`  
**Packages:** pandas, numpy, openpyxl (Excel writer).

### Important finding about the sales file
The sales export is a **single aggregated list per item** (columns: net order intake I/II as amount, item quantities, available stock). It has **no date, no order/transaction ID and no order count**. Therefore the two requested periods - **calendar year 2025** and **Nov 2025 - Oct 2026** - **cannot be calculated** from this file; nothing is estimated or guessed. The notebook reports this explicitly and instead produces *undated, all-available-data* summaries. If you later provide a transaction-level file with a date column, set `SALES_DATE_COL` and the period logic below works unchanged.

```python
from pathlib import Path
import re, warnings, pandas as pd, numpy as np
warnings.filterwarnings('ignore')
pd.set_option('display.width', 200, 'display.max_colwidth', 80, 'display.float_format', '{:,.2f}'.format)

# ---------------- CONFIG ----------------
DATA_DIRS = [Path('SRC'), Path('.'), Path('..')/'SRC', Path('/mnt/data')]
OUT = Path('RES'); OUT.mkdir(exist_ok=True)
SALES_FILE, PRODUCT_FILE = '1. Group_ Orders & sales.csv', 'dwh_present_select.csv'
SALES_DATE_COL = None            # set to a date column name if a dated transaction file is ever supplied (the current file has none)
ORDER_ID_COL = None              # likewise for an order / transaction id column
PERIODS = {'2025_calendar_year': ('2025-01-01', '2025-12-31'),
           '2025-11_to_2026-10': ('2025-11-01', '2026-10-31')}
# ASSUMPTIONS on column meaning (not documented in the file; change if you know better):
#  'Net order intake I (group)'  = order value in EUR before the second adjustment      -> noi1_amount
#  'Order item qty (group)'      = ordered units belonging to NOI I                      -> qty1
#  'Net order intake II (group)' = order value after further adjustments (e.g. cancellations/returns) -> noi2_amount
#  'NOI II item qty (group)'     = units belonging to NOI II                             -> qty2
#  'Stock qty - available'       = current stock, not a sales measure                    -> stock_qty
# The headline 'revenue' used below is NOI II (net); NOI I is kept alongside.
HEADLINE = 'noi2'
```

## 1. Load sales (read-only) and parse German number format

```python
def find_file(name):
    for d in DATA_DIRS:
        if (d/name).exists(): return d/name
    raise FileNotFoundError(name)
def read_csv_robust(path, **kw):
    for enc in ('utf-8-sig', 'cp1252', 'latin1'):
        try: return pd.read_csv(path, encoding=enc, **kw)
        except UnicodeDecodeError: continue
dq = []
def log(check, severity, detail, item_id=None, name=None, subcat=None):
    dq.append(dict(notebook='03', check=check, severity=severity, ItemId=item_id, product_name=name, subcategory=subcat, detail=detail))

raw = read_csv_robust(find_file(SALES_FILE), dtype=str, keep_default_na=False)
raw.columns = ['ItemId', 'sales_name', 'sales_item_type', 'noi1_amount', 'qty1', 'noi2_amount', 'qty2', 'stock_qty']
total_row = raw[raw.ItemId == 'Total'].iloc[0]
s = raw[raw.ItemId != 'Total'].copy()
def parse_de(x):        # '3.403.382' -> 3403382 ; thousands separator '.', decimal ',' (none present)
    x = str(x).strip().replace('.', '').replace(',', '.')
    return pd.to_numeric(x, errors='coerce')
for c in ['noi1_amount', 'qty1', 'noi2_amount', 'qty2', 'stock_qty']:
    s[c] = s[c].map(parse_de)
    bad = s[c].isna().sum()
    if bad: log('sales_unparseable_number', 'high', f'{bad} values in {c} not numeric -> treated as 0')
    s[c] = s[c].fillna(0)
print(s.shape, '| columns:', list(raw.columns))
print('date column present:', SALES_DATE_COL in raw.columns if SALES_DATE_COL else False)
```

## 2. Sales data checks (reconciliation, duplicates, odd values)

```python
# reconciliation against the file's own 'Total' row
rec = []
for c in ['noi1_amount', 'qty1', 'noi2_amount', 'qty2', 'stock_qty']:
    tot = parse_de(total_row[c]); calc = s[c].sum()
    rec.append(dict(measure=c, total_row=tot, sum_of_items=calc, difference=calc - tot, diff_pct=(calc - tot) / tot * 100 if tot else np.nan))
    if abs(calc - tot) > 0.0005 * abs(tot): log('sales_total_row_mismatch', 'medium', f'{c}: sum of item rows {calc:,.0f} vs Total row {tot:,.0f} (diff {calc-tot:,.0f}); possible rounding of amounts in the export or hidden rows')
rec = pd.DataFrame(rec); display(rec)
dup = s[s.ItemId.duplicated(keep=False)]
if len(dup): log('sales_duplicate_item_id', 'high', f'{dup.ItemId.nunique()} ItemIds appear more than once; rows are summed')
neg = s[s.ItemId.str.startswith('-')]
log('sales_negative_item_ids', 'medium', f'{len(neg)} items with negative ItemId (return/manual items, e.g. "(Retournr, ...)"): NOI II {neg.noi2_amount.sum():,.0f}; none exist in product master')
log('sales_negative_values', 'medium', f"{(s.noi2_amount<0).sum()} rows with negative NOI II, {(s.qty2<0).sum()} with negative qty II (returns); kept as-is")
log('sales_amount_without_qty', 'medium', f"{((s.noi2_amount!=0)&(s.qty2==0)).sum()} rows with amount but zero quantity; {((s.noi2_amount==0)&(s.qty2!=0)).sum()} rows with quantity but zero amount")
log('sales_no_date_column', 'high', 'Sales file has no date/period field -> calendar-year-2025 and Nov-2025-Oct-2026 summaries cannot be calculated; undated all-data summaries produced instead')
log('sales_no_order_id', 'high', 'Sales file has no order/transaction ID or order count -> number of orders not available; quantities (item qty) reported instead')
log('sales_noi_i_vs_ii', 'info', f"NOI I total {s.noi1_amount.sum():,.0f} vs NOI II total {s.noi2_amount.sum():,.0f}; difference {(1 - s.noi2_amount.sum()/s.noi1_amount.sum())*100:.1f}% (meaning of I vs II is an assumption)")
log('sales_future_dates', 'info', 'Cannot test for future dates because the file has no dates; the requested window Nov 2025 - Oct 2026 extends beyond any date that could be verified from the data')
s = s.groupby('ItemId', as_index=False).agg(sales_name=('sales_name', 'first'), sales_item_type=('sales_item_type', 'first'), noi1_amount=('noi1_amount', 'sum'), qty1=('qty1', 'sum'),
                                           noi2_amount=('noi2_amount', 'sum'), qty2=('qty2', 'sum'), stock_qty=('stock_qty', 'sum'))
s.head()
```

## 3. Period check - what can and cannot be calculated

```python
def period_status(name, start, end):
    if SALES_DATE_COL is None:
        return dict(period=name, start=start, end=end, status='NOT COMPUTABLE', records_in_period=0,
                    explanation='The sales file contains no date or transaction-level data, so records cannot be assigned to this period. No estimate was made.')
    d = pd.to_datetime(raw[SALES_DATE_COL], errors='coerce')
    inside = d.between(start, end)
    fut = (d > pd.Timestamp.today()).sum()
    return dict(period=name, start=start, end=end, status='COMPUTED' if inside.any() else 'NO RECORDS', records_in_period=int(inside.sum()),
                explanation=f'{int(inside.sum())} dated records in window; {int(fut)} records are dated in the future relative to today; date range in file {d.min()} to {d.max()}')
status = pd.DataFrame([period_status(k, *v) for k, v in PERIODS.items()])
status.to_csv(OUT/'05_sales_period_status.csv', index=False, encoding='utf-8-sig')
# each requested period gets its own clearly named file; while no date data exist it contains only the status (never fake numbers)
for k in PERIODS:
    status[status.period == k].to_csv(OUT/f'05_sales_summary_{k}.csv', index=False, encoding='utf-8-sig')
status
```

## 4. Undated summaries (ALL data in the file, no period filter)
Per product, per configured subcategory and per Shopify category. Measures: net order intake I/II (EUR, assumption), units I/II, average value per unit, NOI I-II gap (share lost between I and II), available stock, stock cover (stock / units II), number of products, number of products with sales, and share of the subcategory.

```python
prod = read_csv_robust(find_file(PRODUCT_FILE), dtype=str).drop_duplicates('ItemId')
assign = read_csv_robust(OUT/'01_product_subcategory_assignments.csv', dtype={'ItemId': str})
sales_only = read_csv_robust(OUT/'01_sales_only_name_candidates.csv', dtype={'ItemId': str})

def measures(g):
    n = len(g)
    r = dict(n_products=n, n_products_with_sales=int((g.qty2 > 0).sum()), noi1_amount=g.noi1_amount.sum(), units_qty1=g.qty1.sum(), noi2_amount=g.noi2_amount.sum(), units_qty2=g.qty2.sum())
    r['avg_value_per_unit_noi1'] = r['noi1_amount'] / r['units_qty1'] if r['units_qty1'] else np.nan
    r['avg_value_per_unit_noi2'] = r['noi2_amount'] / r['units_qty2'] if r['units_qty2'] else np.nan
    r['noi_gap_I_minus_II'] = r['noi1_amount'] - r['noi2_amount']
    r['noi_gap_pct_of_I'] = r['noi_gap_I_minus_II'] / r['noi1_amount'] * 100 if r['noi1_amount'] else np.nan
    r['stock_qty_available'] = g.stock_qty.sum()
    r['stock_cover_units_per_unit_sold'] = r['stock_qty_available'] / r['units_qty2'] if r['units_qty2'] > 0 else np.nan
    r['orders_count'] = 'n/a (no order id in file)'
    return pd.Series(r)

# --- product level: ALL master products (assigned flag + subcategory) ---
P = prod[['ItemId','Name','ManufacturerId','ShopifyTopCategory','ShopifyMainCategory','ShopifySubCategory']].merge(
        assign[['ItemId','SKU','config_subcategory','assignment_confidence']], on='ItemId', how='left').merge(s.drop(columns='sales_name'), on='ItemId', how='left')
P['has_sales_record'] = P.qty1.notna()
for c in ['noi1_amount','qty1','noi2_amount','qty2','stock_qty']: P[c] = P[c].fillna(0)
P['avg_value_per_unit_noi2'] = np.where(P.qty2 > 0, P.noi2_amount / P.qty2.where(P.qty2 > 0), np.nan)
P['noi_gap_I_minus_II'] = P.noi1_amount - P.noi2_amount
P['period'] = 'ALL DATA (undated)'
P['orders_count'] = 'n/a'
P = P.rename(columns={'config_subcategory': 'configured_subcategory', 'qty1': 'units_qty1', 'qty2': 'units_qty2', 'stock_qty': 'stock_qty_available'})
P = P.sort_values('noi2_amount', ascending=False)
P_assigned = P[P.configured_subcategory.notna()].copy()
P_assigned['share_of_subcategory_noi2_pct'] = P_assigned.noi2_amount / P_assigned.groupby('configured_subcategory').noi2_amount.transform('sum') * 100
P.to_csv(OUT/'05_sales_by_product_all_master_products_undated.csv', index=False, encoding='utf-8-sig')
P_assigned.to_csv(OUT/'05_sales_by_product_assigned_undated.csv', index=False, encoding='utf-8-sig')

# --- configured subcategory level ---
sub = P_assigned.rename(columns={'units_qty1': 'qty1', 'units_qty2': 'qty2', 'stock_qty_available': 'stock_qty'}).groupby('configured_subcategory').apply(measures).reset_index()
sub['period'] = 'ALL DATA (undated)'
# supplementary info: sales-only items (not in master) whose NAME matches the subcategory - NOT included in the figures above
so = sales_only.merge(s[['ItemId','noi2_amount','qty2']], on='ItemId', how='left').groupby('candidate_subcategory').agg(
        sales_only_name_match_items=('ItemId', 'nunique'), sales_only_name_match_noi2=('noi2_amount', 'sum'), sales_only_name_match_units2=('qty2', 'sum')).reset_index()
sub = sub.merge(so, left_on='configured_subcategory', right_on='candidate_subcategory', how='left').drop(columns='candidate_subcategory')
sub.to_csv(OUT/'05_sales_by_subcategory_undated.csv', index=False, encoding='utf-8-sig')
display(sub)

# --- category level (Shopify top / main / sub) for the whole master ---
Pn = P.rename(columns={'units_qty1': 'qty1', 'units_qty2': 'qty2', 'stock_qty_available': 'stock_qty'})
cat_tables = {}
for lvl in ['ShopifyTopCategory', 'ShopifyMainCategory', 'ShopifySubCategory']:
    keys = {'ShopifyTopCategory': ['ShopifyTopCategory'], 'ShopifyMainCategory': ['ShopifyTopCategory', 'ShopifyMainCategory'], 'ShopifySubCategory': ['ShopifyTopCategory', 'ShopifyMainCategory', 'ShopifySubCategory']}[lvl]
    t = Pn.groupby(keys, dropna=False).apply(measures).reset_index().sort_values('noi2_amount', ascending=False)
    t['period'] = 'ALL DATA (undated)'
    cat_tables[lvl] = t
    t.to_csv(OUT/f'05_sales_by_{lvl}_undated.csv', index=False, encoding='utf-8-sig')
# categories of the six configured subcategories ('relevant categories')
rel = cat_tables['ShopifySubCategory'][cat_tables['ShopifySubCategory'].ShopifySubCategory.isin(sum([['Vasen'], ['Esstische'], ['Loungesessel'], ['Regalsysteme'], ['Akkuleuchten'], ['Tassen', 'Becher', 'Espressotassen']], []))]
rel.to_csv(OUT/'05_sales_by_relevant_categories_undated.csv', index=False, encoding='utf-8-sig')
cat_tables['ShopifyMainCategory'].head(10)
```

```python
# sales coverage: how much of the sales file belongs to master products?
in_master = s.ItemId.isin(prod.ItemId)
cov = pd.DataFrame({'group': ['items in product master', 'items NOT in master (e.g. delisted, return items)'], 'items': [in_master.sum(), (~in_master).sum()],
                    'noi2_amount': [s[in_master].noi2_amount.sum(), s[~in_master].noi2_amount.sum()], 'units_qty2': [s[in_master].qty2.sum(), s[~in_master].qty2.sum()]})
cov['share_of_noi2_pct'] = cov.noi2_amount / cov.noi2_amount.sum() * 100
cov.to_csv(OUT/'05_sales_coverage_master_vs_unmatched.csv', index=False, encoding='utf-8-sig')
log('sales_unmatched_to_master', 'high', f"{(~in_master).sum()} sales items ({cov.share_of_noi2_pct.iloc[1]:.1f}% of NOI II) have no product master record; they cannot be assigned to any category and are excluded from category/subcategory sums (see 05_sales_coverage_master_vs_unmatched.csv)")
# sales type of assigned products (e.g. delisted items still included)
t = P_assigned
print(t.groupby(['configured_subcategory','sales_item_type'], dropna=False).size().unstack(fill_value=0))
for sc, g in P_assigned.groupby('configured_subcategory'):
    nz = (~g.has_sales_record).sum(); z = ((g.units_qty2 == 0) & g.has_sales_record).sum()
    log('assigned_products_without_sales', 'info', f'{sc}: {z} assigned products have zero NOI II units and {nz} no sales row', subcat=sc)
cov
```

## 5. Consolidated data-quality and exception report

```python
d1 = read_csv_robust(OUT/'dq_nb01.csv', dtype={'ItemId': str}); d2 = read_csv_robust(OUT/'dq_nb02.csv', dtype={'ItemId': str})
dqr = pd.concat([d1, d2, pd.DataFrame(dq)], ignore_index=True)
dqr.insert(0, 'issue_id', range(1, len(dqr) + 1))
# row-level exceptions that must not be lost: assignments and attribute values needing review
iav = read_csv_robust(OUT/'02_item_attribute_value.csv', dtype={'ItemId': str})
rev1 = read_csv_robust(OUT/'01_assignment_review.csv', dtype={'ItemId': str})
extra = pd.concat([
    rev1.assign(notebook='01', check=rev1.review_type, severity='medium', product_name=rev1.Name, subcategory=rev1.candidate_subcategory, detail=rev1.explanation)[['notebook','check','severity','ItemId','product_name','subcategory','detail']],
    iav[iav.review_status.isin(['needs_review'])].assign(notebook='02', check='attribute_value_needs_review', severity='low', detail=iav.attribute + ' = ' + iav.selected_value + ' | ' + iav.source + ' | ' + iav.evidence)[['notebook','check','severity','ItemId','product_name','subcategory','detail']],
], ignore_index=True)
extra.insert(0, 'issue_id', range(len(dqr) + 1, len(dqr) + len(extra) + 1))
exceptions_all = pd.concat([dqr, extra], ignore_index=True)
exceptions_all.to_csv(OUT/'06_data_quality_exception_report.csv', index=False, encoding='utf-8-sig')
dqr.to_csv(OUT/'06_data_quality_summary_checks.csv', index=False, encoding='utf-8-sig')
print(len(dqr), 'summary checks;', len(extra), 'row-level exceptions')
dqr[dqr.severity.isin(['high', 'medium'])][['issue_id','notebook','check','severity','detail']]
```

## 6. Bundle everything into one Excel workbook

```python
readme = pd.DataFrame({'topic': ['Purpose', 'Inputs', 'Periods', 'Sales measures', 'Confidence', 'Row ids', 'Attribute source order', 'Orders'],
  'note': ['Product-to-subcategory assignment, one value per attribute, evidence and review status, sales summaries and exception report.',
           'Subcategories_Matrix_attributes.csv, dwh_present_select.csv, 1. Group_ Orders & sales.csv (unchanged).',
           'Calendar 2025 and Nov 2025-Oct 2026 could NOT be computed: sales file has no date field. See sheet Period_Status. Sales sheets are ALL DATA (undated).',
           'NOI I/II = net order intake (EUR assumed); units = item qty; stock is current stock, not sales.',
           'high = stated in data; medium = derived/heuristic; low = weak signal or default guess -> needs_review.',
           'ItemId = product ID = SKU (no separate SKU column). No order/transaction id in the data.',
           'Product master text/dimensions -> brand/keyword heuristics -> (optional) connox.de -> default guess marked low.',
           'Number of orders not available; only item quantities.']})
sheets = {'README': readme, 'Assignments': assign, 'Item_Attribute_Value': iav, 'Attr_Review_Queue': iav[iav.review_status.isin(['needs_review', 'assignment_needs_review'])],
          'Assignment_Review': rev1, 'Period_Status': status, 'Sales_Subcategory': sub, 'Sales_Relevant_Categories': rel, 'Sales_Main_Category': cat_tables['ShopifyMainCategory'],
          'Sales_Product_Assigned': P_assigned, 'Sales_Reconciliation': rec, 'Sales_Coverage': cov, 'DQ_Summary_Checks': dqr}
with pd.ExcelWriter(OUT/'Product_Attribute_Report.xlsx', engine='openpyxl') as xw:
    for n, df in sheets.items():
        df.to_excel(xw, sheet_name=n[:31], index=False)
        ws = xw.sheets[n[:31]]; ws.freeze_panes = 'A2'; ws.auto_filter.ref = ws.dimensions
        for i, col in enumerate(df.columns, 1):
            ws.column_dimensions[ws.cell(1, i).column_letter].width = min(60, max(10, int(df[col].astype(str).str.len().quantile(0.9)) + 2 if len(df) else 12))
print('workbook written;', {k: len(v) for k, v in sheets.items()})
sorted(p.name for p in OUT.iterdir())
```
